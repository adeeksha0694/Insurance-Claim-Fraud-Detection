```python
import pandas as pd

# load the original dataset
df = pd.read_csv("../data/raw/insurance_claim_fraud_dataset.csv")

print("dataset shape:", df.shape)
print("\nfirst five rows:")
print(df.head())
```

```python
# inspect dataset structure

print("rows:", df.shape[0])
print("columns:", df.shape[1])

print("\nduplicate claim ids:", df["claim_id"].duplicated().sum())

print("\nduplicate rows:", df.duplicated().sum())

print("\nmissing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\ndata types:")
print(df.dtypes.value_counts())
```

```python
# inspect date columns and column names

print("date-related columns:")
print([
    col for col in df.columns
    if "date" in col.lower() or "time" in col.lower()
])

print("\nall column names:")
for i, col in enumerate(df.columns, start=1):
    print(i, col)
```

```python
# convert date columns

date_columns = [
    "claim_date",
    "incident_date",
    "policy_start_date"
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")

print("date data types:")
print(df[date_columns].dtypes)

print("\nmissing dates after conversion:")
print(df[date_columns].isnull().sum())

print("\ndataset shape:", df.shape)
```

```python
# inspect original columns used in eda

print("eda dataset shape:", df.shape)

print("\neda columns:")
print(df.columns.tolist())
```

```python
# create financial exception flags

df["claim_exceeds_coverage"] = (
    df["claim_amount"] > df["coverage_amount"]
)

df["payout_exceeds_approved"] = (
    df["payout_amount"] > df["approved_amount"]
)

df["payout_exceeds_coverage"] = (
    df["payout_amount"] > df["coverage_amount"]
)

print("financial exception summary:")

print(
    df[
        [
            "claim_exceeds_coverage",
            "payout_exceeds_approved",
            "payout_exceeds_coverage"
        ]
    ].sum()
)

```

```python
# validate prepared dataset

print("dataset shape:", df.shape)

print("\nduplicate claim ids:", df["claim_id"].duplicated().sum())

print("\nduplicate rows:", df.duplicated().sum())

print("\nmissing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nfinancial flag data types:")
print(
    df[
        [
            "claim_exceeds_coverage",
            "payout_exceeds_approved",
            "payout_exceeds_coverage"
        ]
    ].dtypes
)
```

```python
from pathlib import Path

# create the processed data folder
output_folder = Path("../data/processed")
output_folder.mkdir(parents=True, exist_ok=True)

# export the prepared dataset
output_file = output_folder / "insurance_claim_fraud_powerbi.csv"
df.to_csv(output_file, index=False)

print("export completed successfully")
print("file path:", output_file.resolve())
print("file size:", round(output_file.stat().st_size / (1024 * 1024), 2), "MB")
```

```python
# reload the exported dataset

df_check = pd.read_csv(output_file)

print("exported dataset shape:", df_check.shape)

print("\nduplicate claim ids:", df_check["claim_id"].duplicated().sum())

print("\nmissing vehicle damage values:",
      df_check["vehicle_damage"].isnull().sum())

print("\nfinancial exception counts:")
print(
    df_check[
        [
            "claim_exceeds_coverage",
            "payout_exceeds_approved",
            "payout_exceeds_coverage"
        ]
    ].sum()
)

print("\nexported columns:")
print(df_check.columns.tolist())
```
