```python
import pandas as pd

df = pd.read_csv("../data/raw/insurance_claim_fraud_dataset.csv")
```

```python
# checking missing value

missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
print(missing)
```

```python
# finding duplicate value

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# checking date pattern

date_columns = [
    "claim_date",
    "incident_date",
    "policy_start_date"
]

for col in date_columns:
    print(f"\n{col}")
    print(df[col].head(5).to_list())
```

```python
# converting date col from string to datetime

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")

df[date_columns].info()
```

```python
# checking id uniqueness

id_columns = ["claim_id", "policy_id", "customer_id"]

for col in id_columns:
    print(f"{col}:")
    print("Unique values:", df[col].nunique())
    print("Duplicate entries:", df[col].duplicated().sum())
    print()
```

```python
# checking fraudulent

print(df["fraudulent"].value_counts(dropna=False))
```

```python
# verifying fraudulent values

print("Unique fraud labels:", df["fraudulent"].unique())
print("Missing fraud labels:", df["fraudulent"].isnull().sum())
```

```python
# validating numeric values

df.describe().T
```

```python
# verifying negative values

check_columns = [
    "claim_amount",
    "approved_amount",
    "deductible",
    "claim_duration_days",
    "age",
    "annual_income",
    "credit_score",
    "previous_claims",
    "premium_amount",
    "coverage_amount",
    "witnesses",
    "injuries",
    "vehicle_age",
    "mileage",
    "vehicle_value",
    "repair_cost",
    "medical_cost",
    "towing_cost",
    "rental_vehicle_cost",
    "payout_amount"
]

for col in check_columns:
    negative_count = (df[col] < 0).sum()

    if negative_count > 0:
        print(f"{col}: {negative_count} negative values")
```

```python
# validating claim amt and approvals

invalid_approved = df["approved_amount"] > df["claim_amount"]

print("Approved amount exceeds claim amount:", invalid_approved.sum())
```

```python
# validating payout amt

invalid_payout = df["payout_amount"] > df["approved_amount"]

print("Payout exceeds approved amount:", invalid_payout.sum())
```

```python
# checking diff b/w payout and aproved amt

invalid_payout = df["payout_amount"] > df["approved_amount"]

difference = (
    df.loc[invalid_payout, "payout_amount"]
    - df.loc[invalid_payout, "approved_amount"]
)

print("Records affected:", len(difference))
print("Smallest excess:", round(difference.min(), 2))
print("Largest excess:", round(difference.max(), 2))
print("Average excess:", round(difference.mean(), 2))
```

```python
#checking whether these records are denied claims

invalid_payout = df["payout_amount"] > df["approved_amount"]

print(
    df.loc[invalid_payout, "claim_status"]
    .value_counts()
)
```

```python
# checking relationship b/w fraud labels

invalid_payout = df["payout_amount"] > df["approved_amount"]

print(
    df.loc[invalid_payout, "fraudulent"]
    .value_counts()
)
```

```python
# checking whether approved amt is zero

invalid_payout = df["payout_amount"] > df["approved_amount"]

print(
    df.loc[invalid_payout, "approved_amount"]
    .describe()
)
```

```python
# Checking the size of the payout excess

invalid_payout = df["payout_amount"] > df["approved_amount"]

difference = (
    df.loc[invalid_payout, "payout_amount"]
    - df.loc[invalid_payout, "approved_amount"]
)

print("Excess above 1,000:", (difference > 1000).sum())
print("Excess above 2,500:", (difference > 2500).sum())
print("Excess above 5,000:", (difference > 5000).sum())
```

```python
#### checking claim dates ####

#checking whether claims submitterd before incidents

invalid_claim_dates = df["claim_date"] < df["incident_date"]

print("Claims submitted before incident:", invalid_claim_dates.sum())
```

```python
#checking whether incidents happen before policy coverage strt date

invalid_policy_dates = df["incident_date"] < df["policy_start_date"]

print("Incidents before policy start:", invalid_policy_dates.sum())
```

```python
# check duration values are consistent with dates

date_difference = (
    df["claim_date"] - df["incident_date"]
).dt.days

print("Exact matches:", (date_difference == df["claim_duration_days"]).sum())
print("Mismatches:", (date_difference != df["claim_duration_days"]).sum())
```

```python
# inspecting mismatched duration dates 

date_difference = (
    df["claim_date"] - df["incident_date"]
).dt.days

mismatches = date_difference != df["claim_duration_days"]

print(
    df.loc[
        mismatches,
        [
            "incident_date",
            "claim_date",
            "claim_duration_days"
        ]
    ].head()
)
```

```python
# checking vehicle age matches year

expected_vehicle_age = (
    df["claim_date"].dt.year - df["vehicle_year"]
)

mismatch_vehicle_age = (
    df["vehicle_age"] != expected_vehicle_age
)

print("Vehicle age mismatches:", mismatch_vehicle_age.sum())
```

```python
# inspecting 5 mismatched records

expected_vehicle_age = (
    df["claim_date"].dt.year - df["vehicle_year"]
)

mismatch_vehicle_age = (
    df["vehicle_age"] != expected_vehicle_age
)

print(
    df.loc[
        mismatch_vehicle_age,
        [
            "claim_date",
            "vehicle_year",
            "vehicle_age"
        ]
    ].head()
)
```

```python
# checking vehicles whuch has manufacturing year after the clai, year

future_vehicle_year = (
    df["vehicle_year"] > df["claim_date"].dt.year
)

print("Vehicle year after claim year:", future_vehicle_year.sum())
```

```python
# checking categorical values

print(df["vehicle_damage"].value_counts(dropna=False))
```

```python
#checking whether missing vehicle damage associated with fraud

pd.crosstab(
    df["vehicle_damage"].isna(),
    df["fraudulent"],
    normalize="index"
).mul(100).round(2)
```

```python
#Checking whether missing vehicle damage varies by claim status
```

```python
pd.crosstab(
    df["claim_status"],
    df["vehicle_damage"].isna(),
    margins=True
)
```

```python
# Calculating missing-value percentages by claim status

missing_by_status = pd.crosstab(
    df["claim_status"],
    df["vehicle_damage"].isna(),
    normalize="index"
).mul(100).round(2)

missing_by_status.columns = [
    "Recorded (%)",
    "Missing (%)"
]

print(missing_by_status)
```

```python
# validating categorical labels in claim_status

print("Unique claim statuses:")
print(df["claim_status"].value_counts(dropna=False))

print("\nNumber of unique statuses:")
print(df["claim_status"].nunique(dropna=False))
```

```python
# Validating all categorical columns together

categorical_columns = df.select_dtypes(
    include=["object", "str", "category", "string"]
).columns

for col in categorical_columns:
    print(f"\n--- {col} ---")
    print("Unique values:", df[col].nunique(dropna=True))
    print("Missing values:", df[col].isna().sum())

    # Checking for leading whitespace
    if df[col].dtype == "object" or str(df[col].dtype) in ["str", "string"]:
        text_values = df[col].dropna().astype(str)

        whitespace_count = (
            text_values != text_values.str.strip()
        ).sum()

        print("Values with extra whitespace:", whitespace_count)

        # Checking for values differing only by capitalization
        case_variants = (
            text_values.str.strip()
            .groupby(text_values.str.strip().str.lower())
            .nunique()
        )

        print(
            "Groups with capitalization variations:",
            (case_variants > 1).sum()
        )

    print("Value counts:")
    print(df[col].value_counts(dropna=False).head(10))
```

```python
# Consolidated numeric and logical validation

print("NUMERIC RANGE CHECKS")
print("-" * 50)

# Values that should normally be positive
positive_columns = [
    "claim_amount",
    "age",
    "annual_income",
    "credit_score",
    "premium_amount",
    "coverage_amount",
    "vehicle_value"
]

for col in positive_columns:
    print(f"{col}: values <= 0 = {(df[col] <= 0).sum()}")


print("\nBOOLEAN LABEL VALIDATION")
print("-" * 50)

boolean_columns = [
    "policy_active",
    "police_report",
    "hospital_visit",
    "property_damage",
    "suspicious_documents",
    "claim_submitted_at_night",
    "multiple_claims_last_year",
    "delayed_reporting",
    "inconsistent_statements",
    "gps_location_match",
    "repair_shop_verified",
    "identity_verified",
    "total_loss",
    "fraudulent"
]

for col in boolean_columns:
    values = set(df[col].dropna().unique())
    print(f"{col}: {values}")


print("\nVEHICLE CONSISTENCY")
print("-" * 50)

print(
    "Vehicle age below zero:",
    (df["vehicle_age"] < 0).sum()
)

print(
    "Vehicle year after claim year:",
    (df["vehicle_year"] > df["claim_date"].dt.year).sum()
)

print(
    "Vehicle age above 50:",
    (df["vehicle_age"] > 50).sum()
)


print("\nCOVERAGE AND CLAIM AMOUNTS")
print("-" * 50)

print(
    "Claims exceeding coverage:",
    (df["claim_amount"] > df["coverage_amount"]).sum()
)

print(
    "Approved amount above claim amount:",
    (df["approved_amount"] > df["claim_amount"]).sum()
)


print("\nCOST COMPONENTS")
print("-" * 50)

cost_columns = [
    "repair_cost",
    "medical_cost",
    "towing_cost",
    "rental_vehicle_cost",
    "payout_amount"
]

for col in cost_columns:
    print(f"{col}: zero values = {(df[col] == 0).sum()}")


print("\nCLAIM HISTORY")
print("-" * 50)

print(
    "Previous claims below zero:",
    (df["previous_claims"] < 0).sum()
)

print(
    "Years with company below zero:",
    (df["years_with_company"] < 0).sum()
)

print(
    "Policy duration below 1 year:",
    (df["policy_duration_years"] < 1).sum()
)
```
